```python
!pip install seaborn
import warnings
warnings.filterwarnings("ignore", category=Warning)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("data/preprocessed_songs.csv")
genre_cols = df.columns[7:]
df = df[df[genre_cols].sum(axis=1) > 0]
df = df.reset_index(drop=True)
df.head()
```

```python
dfclean = df.drop(columns=["Unnamed: 0", "artist", "track", "tags"])
dfclean.head()
```

```python
#Standardization
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(dfclean)
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(n_neighbors=10, metric='cosine')
model.fit(X_scaled)
```

```python
#Test out the model here with one song 
song_index = 0  

distances, indices = model.kneighbors(X_scaled[song_index].reshape(1, -1))

recommended_songs = df.iloc[indices[0]]
print(recommended_songs[["artist", "track"]])

#Test similarities of active genres from our song and reccs
original = df.iloc[song_index][genre_cols]
recs = df.iloc[indices[0]][genre_cols]
active_genres = original[original == 1].index
if len(active_genres) == 0:
    consistency_score = 0
else:
    matches = recs[active_genres].sum(axis=1)
    consistency_score = matches.mean() / len(active_genres)

print("Similarity Score:", consistency_score)
```

```python
#Full functions for every single song 
def recommend (track_name,k=10):
    index = get_index(track_name)
    if index is None:
        return None, 0
    recommended_songs, indices = get_reccs(index)
    score = get_similarity_score(index,indices)
    return score
def get_index(track_name):
    matches = df[(df["track"].str.lower() == track_name.lower())]
    if matches.empty:
        print("Track name not found. Try again.")
        return None
    return matches.index[0]

def get_reccs(song_index):
    distances, indices = model.kneighbors(X_scaled[song_index].reshape(1, -1))
    indices = indices[0][1:]
    recommended_songs = df.iloc[indices]
    print(recommended_songs[["artist", "track"]])
    return recommended_songs, indices

#Overall overlap
def get_similarity_score(song_index,indices):
    original = df.iloc[song_index][genre_cols]
    recs = df.iloc[indices][genre_cols]
    
    active_genres = original[original == 1].index
    matches = recs[active_genres].sum(axis=1)
    consistency_score = matches.mean() / len(active_genres)
    
    print("Similarity Score:", consistency_score)
    return consistency_score
```

```python
def precision_at_k(song_index, k=10):

    # original song genres
    original = df.iloc[song_index][genre_cols]
    active_genres = original[original == 1].index

    if len(active_genres) == 0:
        return 0

    # get recommendations from KNN
    distances, indices = model.kneighbors(
        X_scaled[song_index].reshape(1, -1),
        n_neighbors=k+1
    )

    rec_indices = indices[0][1:]  # remove self
    recs = df.iloc[rec_indices]

    # check relevance (shares at least 1 genre)
    relevant_count = 0

    for i in rec_indices:
        rec_genres = df.iloc[i][genre_cols]

        if (rec_genres[active_genres].sum() > 0):
            relevant_count += 1

    return relevant_count / k
```

```python
song_index = 0
print("Precision@K:", precision_at_k(song_index, k=10))
```

```python
total = 0
n = len(df)

for i in range(n):
    total += precision_at_k(i, k=10)

print("Average Precision@10:", total / n)
```

```python
def evaluate_model(n_neighbors=10, metric='cosine'):
    
    model = NearestNeighbors(n_neighbors=n_neighbors, metric=metric)
    model.fit(X_scaled)
    
    total = 0
    n = len(df)
    
    for i in range(n):
        distances, indices = model.kneighbors(
            X_scaled[i].reshape(1, -1)
        )
        
        rec_indices = indices[0][1:]
        
        original = df.iloc[i][genre_cols]
        active_genres = original[original == 1].index
        
        if len(active_genres) == 0:
            continue
        
        recs = df.iloc[rec_indices][genre_cols]
        matches = recs[active_genres].sum(axis=1)
        
        score = matches.mean() / len(active_genres)
        total += score
    
    return total / n
```

```python
neighbors_list = [10, 15, 20]
metrics_list = ['cosine', 'euclidean']

results = []

for k in neighbors_list:
    for m in metrics_list:
        score = evaluate_model(n_neighbors=k, metric=m)
        results.append((k, m, score))
        print(f"k={k}, metric={m}, score={score}")
```

```python
import matplotlib.pyplot as plt

cosine_scores = []
euclidean_scores = []
ks = []

for k, metric, score in results:
    if metric == "cosine":
        cosine_scores.append(score)
    else:
        euclidean_scores.append(score)
    
    if k not in ks:
        ks.append(k)

plt.figure()

plt.plot(ks, cosine_scores, label="Cosine")
plt.plot(ks, euclidean_scores, label="Euclidean")

plt.title("KNN Hyperparameter Tuning")
plt.xlabel("n_neighbors")
plt.ylabel("Average Similarity Score")
plt.legend()

plt.show()
```
