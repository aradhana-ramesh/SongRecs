```python
import pandas as pd





df1 = pd.read_csv("data/songs_with_tags.csv")
df2 = pd.read_csv("data/Ramesh_songs(in).csv")


df2 = df2.rename(columns = {"song_tags": "tags"})
df2 =df2.drop(columns = ['artist_tags'])
df2 = df2.dropna()
df1 = df1.dropna()
result_df = pd.concat([df1, df2], ignore_index=True)
result_df = result_df.sort_values(by=['artist', 'track'])
```

```python
result_df
```

```python
result_df = result_df.groupby(["artist", "track"], as_index=False).agg({"tags": ", ".join, "listeners": "max", "playcount": "max"})
```

```python
result_df['ratio'] = result_df['listeners']/result_df['playcount']
```

```python
result_df.to_csv("data/final_songs_with_tags.csv", index=False)
```
