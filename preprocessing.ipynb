```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from collections import Counter
import re
from sklearn.preprocessing import StandardScaler


df = pd.read_csv("data/Final_Cleaned_Songs_List.csv")


df.head()
df.info()
```

```python
df_filtered = df[~df['track'].str.contains("- Live")]


df_filtered.loc[:,'tags'] = df_filtered['tags'].apply(lambda x: [genre.strip() for genre in x.split(',')])
df_filtered.info()
```

```python
all_tags = []
for r in df_filtered.itertuples():
    all_tags.extend(r.tags)
all_tags
```

```python
bad_genres=['wsum 91.7 fm madison','bts', 'love at first listen', 'female vocalists', 'singer-songwriter', '70s', '80s', '90s', '2010s', 'beautiful']

all_tags = [item.strip() for item in all_tags if item.strip().lower() not in bad_genres]
```

```python
from collections import Counter




counter = Counter(all_tags)

# for word in all_tags:
#     tags = word.split(", ")
#     counter.update(tags)

top50 = counter.most_common(50)


# print(top50)


top_words = [word for word, count in  counter.most_common(50)]
```

```python
top_words
```

```python
df_filtered.loc[:,'tags'] = df_filtered['tags'].apply(
    lambda x: [item for item in x if item.strip() in top_words]
)

print(df_filtered)
```

```python
df_filtered.loc[:,'plct/lstn'] = df_filtered['playcount']/df_filtered['listeners']





```

```python

df_filtered.loc[:,'tags'] = df_filtered['tags'].apply(
    lambda x: [item for item in x if item.strip() in top_words]
)
df_filtered = df_filtered[df_filtered['tags'].apply(lambda x: len(x) > 0)]

print(df_filtered)
df_filtered.head()
```

```python
exploded = df_filtered.explode('tags')

encoded = pd.get_dummies(exploded['tags'])
```

```python
encoded_final = encoded.groupby(exploded.index).sum()
encoded_final
```

```python
df_filtered_encoded = df_filtered.join(encoded_final, how='left')
df_filtered_encoded
```

```python
scaler = StandardScaler()
scaler.fit(df_filtered_encoded.select_dtypes(include=['number']))
df_filtered_encoded = scaler.fit_transform(df_filtered_encoded.select_dtypes(include=['number']))
# df_filtered_encoded = scaler.fit_transform(df_filtered_encoded[['listeners', 'playcount','plct/lstn']])
```
